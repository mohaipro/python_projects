```python
"""
Welcome to the mini terminal game "Number Guessing Game" in Python!
Run the function play_game to start the game.
Happy guessing!
"""
# importing the random module to build upton it the (Guessing Game!)
import random

def generate_random_number():
    """Generate a random number between 1 and 10."""
    return random.randint(1, 10)

def play_game():
    """Play an interactive number guessing game."""
    print("\n🎉 Welcome to the Number Guessing Game! 🎉")
    print("Try to guess the secret number between 1 and 10.")
    print("You start with **5 attempts**, and each incorrect guess will reduce your chances.")
    print("Be careful, and **good luck!** 🚀\n")

    attempts = 5
    secret_number = generate_random_number()

    while attempts > 0:
        print(f"\n🔹 Attempts remaining: {attempts}")

        try:
            guess = int(input("Enter your guess (1-10): "))
        except ValueError:
            print("❌ Invalid input! Please enter a **number** between 1 and 10.")
            continue  # Skip to next iteration

        if guess < 1 or guess > 10:
            print("⚠️ Out of range! Please enter a **number** between 1 and 10.")
            continue

        if guess == secret_number:
            print(f"\n✅ **Congratulations!** You guessed the right number: {secret_number}")
            print("🎊 **You won the game!** 🏆")
            break
        else:
            hint = "higher" if guess < secret_number else "lower"
            print(f"❌ Wrong guess! Try a {hint} number.")
            attempts -= 1

    if attempts == 0:
        print(f"\n💀 **Game Over!** The correct number was {secret_number}. Better luck next time!")

# Start the game
play_game()
```
